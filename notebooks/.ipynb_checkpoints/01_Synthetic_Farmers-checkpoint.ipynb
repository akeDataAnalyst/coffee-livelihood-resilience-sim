#### Smallholder Livelihood Simulation for Enveritas  

**Focus**: Realistic synthetic population of 1,000 smallholder coffee farmers.

Key characteristics are based on field studies in Jimma:
- Small coffee plots (mostly 0.3–1.0 ha)
- Low to moderate yields
- High vulnerability to price shocks and debt
- Limited diversification
- Coffee as main cash crop (often 50–70% of income)

This population will be used later in the Agent-Based Model to test interventions (e.g., fertilizer support, micro-loans for diversification, training).

#### Imports

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Reproducibility
np.random.seed(42)

print("Libraries imported successfully.")
```

#### Define Realistic Parameters

#### Key Parameters
- Coffee farm size: mostly 0.3–1.0 ha (average 0.6 ha)
- Total land holding: 1.2–2.8 ha
- Baseline coffee yield: 400–900 kg/ha (green coffee equivalent)
- Debt and savings: many farmers face liquidity issues and limited savings
- Diversification: low for most (coffee dominant)
- Household size: average 5.5–7 persons

```python
n_farmers = 1000

# Coffee-specific farm sizes (hectares) - lognormal to reflect many very small farms
coffee_area_ha = np.random.lognormal(mean=-0.6, sigma=0.55, size=n_farmers)
coffee_area_ha = np.clip(coffee_area_ha, 0.2, 2.5)

# Total land (slightly larger than coffee area)
total_land_ha = coffee_area_ha + np.random.uniform(0.3, 1.5, n_farmers)
total_land_ha = np.clip(total_land_ha, 0.8, 5.0)

# Baseline coffee yield (kg/ha green coffee) - lower end common in smallholder systems
baseline_yield_kg_ha = np.random.normal(650, 180, n_farmers)
baseline_yield_kg_ha = np.clip(baseline_yield_kg_ha, 300, 1200)

# Debt (ETB) - many have moderate debt for inputs or consumption
debt_etb = np.random.exponential(scale=8000, size=n_farmers)
debt_etb = np.clip(debt_etb, 0, 45000)

# Savings (ETB) - generally low
savings_etb = np.random.exponential(scale=2500, size=n_farmers)
savings_etb = np.clip(savings_etb, 0, 20000)

# Diversification level (0-1): low for most coffee-dependent farmers
diversification_level = np.random.beta(a=1.8, b=5.5, size=n_farmers)

# Risk tolerance (0-1)
risk_tolerance = np.random.beta(a=2.5, b=4, size=n_farmers)

# Household size
household_size = np.random.normal(6.2, 1.8, n_farmers).astype(int)
household_size = np.clip(household_size, 3, 12)

# Region fixed as Jimma Zone (for future shock modeling)
region = ['Jimma Zone, Ethiopia'] * n_farmers
```

#### Create the Synthetic Farmer DataFrame

```python
# Create DataFrame
farmers = pd.DataFrame({
    'farmer_id': range(1, n_farmers + 1),
    'coffee_area_ha': coffee_area_ha.round(2),
    'total_land_ha': total_land_ha.round(2),
    'baseline_yield_kg_ha': baseline_yield_kg_ha.round(0).astype(int),
    'debt_etb': debt_etb.round(0).astype(int),
    'savings_etb': savings_etb.round(0).astype(int),
    'diversification_level': diversification_level.round(3),
    'risk_tolerance': risk_tolerance.round(3),
    'household_size': household_size,
    'region': region
})

# Derived realistic columns
farmers['baseline_coffee_production_kg'] = (farmers['coffee_area_ha'] * farmers['baseline_yield_kg_ha']).round(0)
farmers['baseline_coffee_income_etb'] = (farmers['baseline_coffee_production_kg'] * 35).round(0)   
farmers['poverty_risk'] = (farmers['debt_etb'] > farmers['savings_etb'] * 1.8).astype(int)  

print(f"Generated {len(farmers):,} synthetic smallholder coffee farmers from Jimma Zone.")
farmers.head()
```

#### Exploratory Data Analysis

```python
display(farmers.describe().round(2))

print("\nPoverty Risk Distribution:")
print(farmers['poverty_risk'].value_counts(normalize=True).round(3) * 100, "%")
```

#### Visualizations

```python
fig, axes = plt.subplots(2, 3, figsize=(16, 9))

sns.histplot(farmers['coffee_area_ha'], kde=True, bins=30, ax=axes[0,0], color='brown')
axes[0,0].set_title('Coffee Farm Size (ha) - Jimma Zone')
axes[0,0].set_xlabel('Coffee Area (ha)')

sns.histplot(farmers['baseline_yield_kg_ha'], kde=True, bins=30, ax=axes[0,1], color='green')
axes[0,1].set_title('Baseline Coffee Yield (kg/ha)')

sns.histplot(farmers['debt_etb']/1000, kde=True, bins=30, ax=axes[0,2], color='red')
axes[0,2].set_title('Debt (thousand ETB)')

sns.histplot(farmers['savings_etb']/1000, kde=True, bins=30, ax=axes[1,0], color='blue')
axes[1,0].set_title('Savings (thousand ETB)')

sns.histplot(farmers['diversification_level'], kde=True, bins=30, ax=axes[1,1], color='orange')
axes[1,1].set_title('Diversification Level (0-1)')

sns.histplot(farmers['household_size'], bins=range(3,13), ax=axes[1,2], color='purple')
axes[1,2].set_title('Household Size')

plt.tight_layout()
plt.show()
```

#### Save the Population

```python
import os
os.makedirs('data/results', exist_ok=True)

farmers.to_csv('data/results/synthetic_jimma_coffee_farmers.csv', index=False)

print("coffee farmer population saved to:")
print(" data/results/synthetic_jimma_coffee_farmers.csv")
```
