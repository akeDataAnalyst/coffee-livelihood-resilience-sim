#### Loading Agent-Level Data

```python
import sys, os
import pandas as pd
from scipy import stats

sys.path.append(os.path.abspath('..'))
from model import LivelihoodModel

# Run the three models again
results_savings = {}

for scn in ["Control", "Intensification", "Diversification"]:
    model = LivelihoodModel(N=1000, intervention=scn)
    for _ in range(60): model.step()
    results_savings[scn] = [a.savings for a in model.agents]

print("Agent-level distribution data collected.")
```

#### The T-Test

```python
# Compare Intensification to Control
t_stat, p_val = stats.ttest_ind(results_savings["Intensification"], results_savings["Control"])

print(f"--- Statistical Rigor: Intensification vs. Control ---")
print(f"T-Statistic: {t_stat:.4f}")
print(f"P-Value: {p_val:.10f}")

if p_val < 0.05:
    print("Result: REJECT the Null Hypothesis. The improvement is statistically significant.")
else:
    print("Result: FAIL TO REJECT the Null Hypothesis. The improvement might be due to chance.")
```
